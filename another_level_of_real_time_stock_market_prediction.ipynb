```python
import pandas as pd
import yfinance as yf
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from datetime import datetime, timedelta
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
company_symbols = ['AAPL', 'MSFT', 'GOOGL', 'AMZN', 'TSLA','NVDA','JPM','GS','WMT','KO']
end_date = datetime.now()
start_date = end_date - timedelta(days=365)
start_date
```

```python
company_data = {}
for symbol in company_symbols:
    df = yf.download(symbol, start=start_date, end=end_date)
    company_data[symbol] = df
```

```python
company_model={}
for symbol,df in company_data.items():
  df=df.copy()
  df['Target']=df['Close'].shift(-1)
  df.dropna(inplace=True)
  x=df[['Close']]
  y=df['Target']
  x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.2,random_state=42)
  model=LinearRegression()
  model.fit(x_train,y_train)
  company_model[symbol]=model
```

```python
for symbol,model in company_data.items():
   model=company_model[symbol]
   x_test=df[['Close']].tail(1)
   y_test=df['Close'].tail(1)
   y_pred=model.predict(x_test)

   mae=mean_absolute_error(y_test,y_pred)
   mse=mean_squared_error(y_test,y_pred)
   rmse=np.sqrt(mse)

   print("company: ",symbol)
   print("MAE: ",mae)
   print("MSE: ",mse)
   print("RMSE: ",rmse)
   print("*"*100)
```
